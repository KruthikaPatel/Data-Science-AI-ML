```python
import numpy as np
import pandas as pd

np.random.seed(42)

# Generate heavily skewed income data
income_data = np.random.lognormal(mean=10, sigma=1, size=100000)

# Convert to DataFrame
df = pd.DataFrame(income_data, columns=["Income"])

print(df.head())
print("Skewness:", df["Income"].skew())
```

```python
sample_means = []

for i in range(1000):
    sample = np.random.choice(df["Income"], size=30)
    sample_mean = np.mean(sample)
    sample_means.append(sample_mean)

print("First 5 sample means:", sample_means[:5])
```

```python
import matplotlib.pyplot as plt

plt.hist(sample_means, bins=30)
plt.title("Distribution of Sample Means (n = 30)")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.show()
```

```python
print("Original Skewness:", df["Income"].skew())
print("Sample Means Skewness:", pd.Series(sample_means).skew())
```

```python
sample_means = []

for i in range(5000):   # increase simulations
    sample = np.random.choice(df["Income"], size=50)  # increase sample size
    sample_means.append(np.mean(sample))
```

```python
import pandas as pd

print("Original Skewness:", df["Income"].skew())
print("Sample Means Skewness:", pd.Series(sample_means).skew())
```

```python
import numpy as np

mu = df["Income"].mean()
sigma = df["Income"].std()

print("Mean:", mu)
print("Standard Deviation:", sigma)
```

```python
df["z_score"] = (df["Income"] - mu) / sigma

print(df.head())
```

```python
outliers = df[np.abs(df["z_score"]) > 3]

print("Number of Outliers:", len(outliers))
print(outliers.head())
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)

# Normal distribution (heights)
heights = np.random.normal(loc=170, scale=10, size=1000)

df_heights = pd.DataFrame({"Heights": heights})

```

```python
sns.histplot(df_heights["Heights"], kde=True)
plt.title("Normal Distribution - Human Heights")
plt.show()

print("Mean:", df_heights["Heights"].mean())
print("Median:", df_heights["Heights"].median())
```

```python
income = np.random.lognormal(mean=10, sigma=1, size=1000)
df_income = pd.DataFrame({"Income": income})

sns.histplot(df_income["Income"], kde=True)
plt.title("Right-Skewed Distribution - Income")
plt.show()

print("Mean:", df_income["Income"].mean())
print("Median:", df_income["Income"].median())
```

```python
scores = 100 - np.random.exponential(scale=10, size=1000)
df_scores = pd.DataFrame({"Scores": scores})

sns.histplot(df_scores["Scores"], kde=True)
plt.title("Left-Skewed Distribution - Easy Exam")
plt.show()

print("Mean:", df_scores["Scores"].mean())
print("Median:", df_scores["Scores"].median())
```
