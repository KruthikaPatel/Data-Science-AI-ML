# Day21 - Introduction to Machine Learning

Machine Learning is a field of Artificial Intelligence that allows computers to learn patterns from data and make decisions without being explicitly programmed with fixed rules. In traditional programming, developers provide the computer with both the data and a set of rules, and the computer follows those rules to produce results. However, in Machine Learning, instead of writing detailed rules, we provide the system with a large amount of data along with the correct answers, and the algorithm learns the underlying patterns by analyzing the data. Once the model is trained, it can apply what it has learned to new and unseen data to make predictions or classifications. This approach is useful in many real-world applications such as recommendation systems, spam detection, image recognition, and predictive analytics. Machine Learning helps computers improve their performance over time as they are exposed to more data and experiences.

1. Supervised Learning

Definition:
Supervised Learning is a type of Machine Learning where the model is trained using labeled data.
This means the dataset already contains the correct answers (labels), and the model learns to predict those answers.

Example idea:
Input → Data with known output
Model learns → Predicts future outputs.

Example:
If emails are labeled “Spam” or “Not Spam”, the model learns to classify new emails.

Common Tasks

Classification – Predict categories (Spam / Not Spam)

Regression – Predict numbers (House price)

2. Unsupervised Learning

Definition:
Unsupervised Learning is when the model is trained using data without labels.
The goal is to find hidden patterns, structures, or clusters in the data.

Example idea:
Input → Raw data
Model → Finds patterns or groups.

Example:
Grouping customers based on shopping behavior without knowing the groups beforehand.

Common Tasks

Clustering

Association

Dimensionality Reduction

3. Reinforcement Learning

Definition:
Reinforcement Learning is when an agent learns by interacting with an environment and receives rewards or penalties based on its actions.

The model improves by maximizing rewards over time.

Example idea:
Agent → Takes action → Gets reward/penalty → Learns better action.

Example:
A robot learning how to walk by trying different movements and improving.
